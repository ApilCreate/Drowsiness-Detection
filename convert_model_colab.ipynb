```python
import tensorflow as tf
import tensorflowjs as tfjs
from pathlib import Path
model = tf.keras.models.load_model('drowsiness_model.h5', compile=False)
assert model.input_shape[1:] == (64, 64, 1)
assert model.output_shape[-1] == 2
Path('web/model').mkdir(parents=True, exist_ok=True)
tfjs.converters.save_keras_model(model, 'web/model')
```
