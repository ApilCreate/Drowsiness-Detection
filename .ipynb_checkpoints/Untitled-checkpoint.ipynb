Preprocessing the Dataset

```python
import cv2
import numpy as np
import os
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split

# Define dataset path
dataset_path = r"E:\ISMT\Drowsiness_Detection\Drowsiness_Dataset"
categories = ["Open_Eyes", "Close_Eyes"]  # Define class labels
```

```python
# Set the target image size
IMG_SIZE = 64  # Resize all images to 64x64
```

```python
# Lists to store images and labels
data = []
labels = []

# Loop through both classes (Open_Eyes, Close_Eyes)
for category in categories:
    folder_path = os.path.join(dataset_path, category)
    label = categories.index(category)  # Assign label (0 for Open, 1 for Closed)

    # Loop through each image file in the folder
    for img_name in os.listdir(folder_path):
        img_path = os.path.join(folder_path, img_name)
        
        # Read and preprocess the image
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  # Convert to grayscale
        img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))  # Resize to 64x64

        data.append(img)
        labels.append(label)
```

```python
# Convert lists to NumPy arrays
data = np.array(data) / 255.0  # Normalize pixel values (0-1)
labels = np.array(labels)

# Reshape images to match CNN input shape
data = data.reshape(-1, IMG_SIZE, IMG_SIZE, 1)

# Convert labels to categorical (one-hot encoding)
labels = to_categorical(labels, num_classes=2)
```

```python
# Split the dataset into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(data, labels, test_size=0.2, random_state=42)
```

```python
# Print dataset details
print(f"Total images: {len(data)}")
print(f"Training images: {len(X_train)}, Testing images: {len(X_test)}")
print("Dataset preprocessing completed successfully!")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, Input  # Add Input

# Define CNN model architecture
model = Sequential([
    Input(shape=(64, 64, 1)),  # Corrected Input layer
    Conv2D(32, (3,3), activation='relu'),  
    MaxPooling2D(2,2),  

    Conv2D(64, (3,3), activation='relu'),  
    MaxPooling2D(2,2),  

    Flatten(),  
    Dense(128, activation='relu'),  
    Dropout(0.5),  
    Dense(2, activation='softmax')  # Output layer for 2 classes
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
# Print model summary
model.summary()
```

Training the CNN Model

```python
# Train the model
history = model.fit(
    X_train, y_train,  # Training data
    epochs=10,  # Number of training loops
    batch_size=32,  # Number of images processed at once
    validation_data=(X_test, y_test)  # Validation data
)
```

```python
# Save the trained model
model.save("drowsiness_model.h5")
```

```python
print("Model training completed and saved successfully! ")
```

```python
import cv2
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# Load the trained model
model = tf.keras.models.load_model("drowsiness_model.h5")

# Define image size
IMG_SIZE = 64
```

```python

# Paths to test images
open_eye_path = r"E:\ISMT\Drowsiness_Detection\Drowsiness_Dataset\Open_Eyes\s0001_01899_0_0_1_0_0_01.png"
closed_eye_path = r"E:\ISMT\Drowsiness_Detection\Drowsiness_Dataset\Close_Eyes\s0001_00022_0_0_0_0_0_01.png"

def predict_eye_state(image_path):
    """Function to load image, preprocess it, and predict the eye state."""
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)  # Convert to grayscale
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))  # Resize
    img = img / 255.0  # Normalize pixel values
    img = img.reshape(1, IMG_SIZE, IMG_SIZE, 1)  # Reshape for CNN

    # Predict open or closed eyes
    prediction = model.predict(img)
    label = "Open" if np.argmax(prediction) == 0 else "Closed"

    # Show the image with prediction
    plt.imshow(cv2.imread(image_path, cv2.IMREAD_GRAYSCALE), cmap='gray')
    plt.title(f"Predicted: {label}")
    plt.axis('off')
    plt.show()
```

Real-Time Drowsiness Detection with OpenCV

```python
import cv2
import numpy as np
import tensorflow as tf
import time
```

```python
# Load the trained model
model = tf.keras.models.load_model("drowsiness_model.h5")
```

```python
# Define image size
IMG_SIZE = 64
```

```python
# Start video capture (0 for default webcam)
cap = cv2.VideoCapture(0)
```

```python
# Load Haar cascades for face and eye detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Convert frame to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(50, 50))

    for (x, y, w, h) in faces:
        face_roi = gray[y:y+h, x:x+w]  # Region of interest (Face)

        # Detect eyes within the face region
        eyes = eye_cascade.detectMultiScale(face_roi, scaleFactor=1.1, minNeighbors=5, minSize=(20, 20))

        for (ex, ey, ew, eh) in eyes:
            eye_roi = face_roi[ey:ey+eh, ex:ex+ew]  # Extract eye region
            eye_roi = cv2.resize(eye_roi, (IMG_SIZE, IMG_SIZE))  # Resize for CNN
            eye_roi = eye_roi / 255.0  # Normalize pixel values
            eye_roi = eye_roi.reshape(1, IMG_SIZE, IMG_SIZE, 1)  # Reshape for model input

            # Predict eye state
            prediction = model.predict(eye_roi)
            confidence = np.max(prediction) * 100  # Get confidence percentage
            label = "Open" if np.argmax(prediction) == 0 else "Closed"

            # Draw rectangle around eyes
            color = (0, 255, 0) if label == "Open" else (0, 0, 255)
            cv2.rectangle(frame, (x+ex, y+ey), (x+ex+ew, y+ey+eh), color, 2)
            cv2.putText(frame, f"{label} ({confidence:.2f}%)", (x+ex, y+ey-10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2)

            # Alert if eyes are closed
            if label == "Closed":
                cv2.putText(frame, "DROWSINESS ALERT!", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)

    cv2.imshow("Drowsiness Detection", frame)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import tensorflow as tf
import time
import threading
from playsound import playsound

# Load the trained model
model = tf.keras.models.load_model("drowsiness_model.h5")

# Define image size
IMG_SIZE = 64

# Start video capture (0 for default webcam)
cap = cv2.VideoCapture(0)

# Load Haar cascades for face and eye detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")

# Flag to track if alarm is already playing
drowsy_detected = False

# Function to play the alarm in a separate thread
def play_alarm():
    playsound("alarm.wav")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Convert frame to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(50, 50))

    for (x, y, w, h) in faces:
        face_roi = gray[y:y+h, x:x+w]  # Extract face ROI

        # Detect eyes within the face region
        eyes = eye_cascade.detectMultiScale(face_roi, scaleFactor=1.1, minNeighbors=5, minSize=(20, 20))

        eye_closed_count = 0  # Track number of closed eyes detected

        for (ex, ey, ew, eh) in eyes:
            eye_roi = face_roi[ey:ey+eh, ex:ex+ew]  # Extract eye region
            eye_roi = cv2.resize(eye_roi, (IMG_SIZE, IMG_SIZE))  # Resize for CNN
            eye_roi = eye_roi / 255.0  # Normalize pixel values
            eye_roi = eye_roi.reshape(1, IMG_SIZE, IMG_SIZE, 1)  # Reshape for model input

            # Predict eye state
            prediction = model.predict(eye_roi)
            confidence = np.max(prediction) * 100  # Get confidence percentage
            label = "Open" if np.argmax(prediction) == 0 else "Closed"

            # Draw rectangle around eyes
            color = (0, 255, 0) if label == "Open" else (0, 0, 255)
            cv2.rectangle(frame, (x+ex, y+ey), (x+ex+ew, y+ey+eh), color, 2)
            cv2.putText(frame, f"{label} ({confidence:.2f}%)", (x+ex, y+ey-10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2)

            if label == "Closed":
                eye_closed_count += 1

        # Check if both eyes are closed
        if eye_closed_count >= 1:  # If at least one eye is detected as closed
            cv2.putText(frame, "DROWSINESS ALERT!", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)
            if not drowsy_detected:  # Play alarm only if it's not already playing
                drowsy_detected = True
                threading.Thread(target=play_alarm, daemon=True).start()
        else:
            drowsy_detected = False  # Reset the flag if eyes are open

    cv2.imshow("Drowsiness Detection", frame)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
